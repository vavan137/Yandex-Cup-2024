```python
import pandas as pd

df = pd.read_excel('/content/F transactions.xlsx')
df_2 = pd.read_excel('/content/F merchants.xlsx')
```

```python
df = df[['cost','countertype','date','fraudbits','merchantid','model_cost']]
df = df.merge(df_2,how='left',on = 'merchantid')
df = df[df.fraudbits==0]
df.head(1)
```

Нужны расчетные поля: Амнистия и Час (операции)


```python
df['amnesty'] = abs(df['model_cost']-df['cost'])/df['coef']
df['hour']= df['date'].str[10:12]
df.head(1)
```

```python
g_1 = df.groupby('guild')['amnesty'].mean()
g_2 = df[df['countertype']=='buy'].groupby('guild')['date'].count()
g_3 = df[df['countertype']=='sell'].groupby('guild')['date'].count()
#g_2
res = pd.concat([g_1,g_2,g_3],axis =1,keys =['av_amnesty','buy','sell'])
res['eff'] = res['sell']/res['buy']*res['av_amnesty']
```

```python
row = res[res['eff']==res['eff'].max()]
row
```
